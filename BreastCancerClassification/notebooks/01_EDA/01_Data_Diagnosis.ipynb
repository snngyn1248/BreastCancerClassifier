```python
import pandas as pd
import numpy as np

filepath = "wdbc.data"

#wdbc.data has no header row — first row is patient data, not column names
COLUMNS = [
    "id", "diagnosis",
    "radius_mean", "texture_mean", "perimeter_mean", "area_mean",
    "smoothness_mean", "compactness_mean", "concavity_meaan",
    "concave_points_mean", "symmetry_mean", "fractal_dimension_mean",
    "radius_se", "texture_se", "perimeter_se", "area_se",
    "smoothness_se", "compactness_se", "concavity_se",
    "concave_points_se", "symmetry_se", "fractal_dimension_se",
    "radius_worst", "texture_worst", "perimeter_worst", "area_worst",
    "smoothness_worst", "compactness_worst", "concavity_worst",
    "concave_points_worst", "symmetry_worst", "fractal_dimension_worst",
]
df = pd.read_csv(filepath, header=None, names=COLUMNS)

ID_COL = "id"
TARGET_COL = "diagnosis"

#1. Row Breakdown
assert df.shape == (569, 32), f"Expected (569, 32), got {df.shape}"
print(f"Loaded {df.shape[0]} observations, {df.shape[1]} columns")

#2. Column breakdown:
feature_cols = [c for c in df.columns if c not in [ID_COL, TARGET_COL]]
assert len(feature_cols) == 30, f"Expected 30 features, got {len(feature_cols)}"
print(f"✓ 1 ID, 1 diagnosis, {len(feature_cols)} numerical features")

#3. Diagnosis values are only M or B (Malignant or Benign)
print("Diagnosis values", df[TARGET_COL].unique())

#4. No Missing Values
missing = df.isna().sum().sum()
assert missing == 0, f"Found {missing} missing values"  #will print this statement if statement is true
print("✓ No missing values")

#5. No duplicated rows
dup_rows = df.duplicated().sum()
assert dup_rows == 0, f"Found {dup_rows} duplicate rows"
print('✓ No duplicated rows')

#6. No duplicated IDs
dup_ids = df[ID_COL].duplicated().sum()
assert dup_ids == 0, f"Found {dup_ids} duplicate IDs"
print("✓ No duplicated ids")

#7. No infinite values (in numeric columns)
inf_count = np.isinf(df.select_dtypes(include="number")).sum().sum()
assert inf_count == 0, f"Found {inf_count} infinite values"
print("✓ No infinite values")

#8. No constant or empty columns
for col in df.columns:
    assert df[col].nunique() > 1, f"Column {col} is constant"
print("✓ No constant columns")

#9. Diagnosis distribution
counts = df[TARGET_COL].value_counts()
pct = df[TARGET_COL].value_counts(normalize=True) *100 #normalize then multiply by 100
print("\nDiagnosis distribution")
print(counts)
print(pct.round(1))

assert counts["B"] == 357
assert counts["M"] == 212
assert abs(pct["B"] - 62.7) < 0.1
assert abs(pct["M"] - 37.3) < 0.1
print("✓ Distribution matches: 357 B (62.7%), 212 M (37.3%)")
```
